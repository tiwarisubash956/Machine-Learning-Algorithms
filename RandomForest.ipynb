```python
# number of estimator in random forest decision tree
n_estimator=[20,30,60,80,100]

# number of feature to be consider
max_features=['auto','sqrt','log2']

# maximum number of levels in tree
max_depth=[2,8,None]

# number of sample
max_samples=[0.5,0.75,1.0]
```
