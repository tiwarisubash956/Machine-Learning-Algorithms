**Reduction operation**

```python
e=torch.randint(size=(2,3),low=2,high=10,dtype=torch.float)
e
```

```python
# sum of elemnents
sum=torch.sum(e)
sum

# sum along rows
sumr=torch.sum(e,dim=1)
sumr

# sum along columns
sumc=torch.sum(e,dim=0)
sumc
```

```python
# mean
torch.mean(e)

# mean of colums
torch.mean(e,dim=0)

# mean of rows
torch.mean(e,dim=1)
```

```python
# median
torch.median(e)

# median of columns
torch.median(e,dim=0)

# median of rows
torch.median(e,dim=1)
```

```python
# max ,min
print(torch.max(e))
print(torch.min(e))

#
```

```python
# product
torch.prod(e)

# product of columns
torch.prod(e,dim=0)

# product of rows
torch.prod(e,dim=1)
```

```python
# standard deviation
torch.std(e)

# standard deviation of columns
torch.std(e,dim=0)

# standard deviation of rows
torch.std(e,dim=1)
#
```

```python
# varinace
torch.var(e)

# varinace of columns
torch.var(e,dim=0)

# varinace of rows
torch.var(e,dim=1)
```

```python
# max along elements using argmax and minimum using arg min
print(torch.argmax(e))
print(torch.argmin(e))
```
